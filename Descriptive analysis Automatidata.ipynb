```python
# We start by importing pandas and the required functions
import pandas as pd
import numpy as np
# Then we import the designated file
df=pd.read_csv("C:/Users/danca/Downloads/Yellow_Taxi_Trip_Data.csv")
```

```python
# Preview data
df.head()
```

```python
#We do an inital check
#Information ,missing values & basic statistics
df.info()
df.isnull()
```

```python
#Now we standardize some formats to be easily used in coding
df['tpep_dropoff_datetime']=pd.to_datetime(df['tpep_dropoff_datetime'])
df['tpep_pickup_datetime']=pd.to_datetime(df['tpep_pickup_datetime'])
```

```python
df.head()
```

```python
#We select crucial variables that will require a more in-depth analysis.
#These variables will become the core variables for regression analysis when determining fare amount.
numerical_cols=['trip_distance','fare_amount','total_amount','tolls_amount','extra','tip_amount','passenger_count']
desc_stats=df[numerical_cols].aggregate(['mean','std','min','max'])
desc_stats
#Mean
#Useful for a simple starting prediction
#Standard Deviation (volatility)
#High std → harder to predict accurately
#Low std → easier, more stable pricing
#Min & Max (Important for Boundaries)
#Helps define realistic prediction limits
```

```python
#Now, next are variables that don't need descriptive analysis but are still important
#Number of payments per method
payment_counts=df['payment_type'].value_counts()
print(payment_counts)
#frequency per Vendor sold
vendor_sales=df['VendorID'].value_counts()
print(vendor_sales)
```

```python
#Next, we categorise variables in certain scenarios.
#Average Amount of tips per method of payment
df.groupby('payment_type')['tip_amount'].mean()
```

```python
#Average number of tips per passenger carried
df.groupby('passenger_count')['tip_amount'].mean()
```

```python
#Amount collected per method of payment
df.groupby('payment_type')['total_amount'].sum()
```

```python
#Total amount collected by each vendor
df.groupby('VendorID')['total_amount'].sum()
```

```python
#Amount earned per number of passengers
df.groupby('passenger_count')['total_amount'].sum()
```

```python
#Average total amount per each vendor
df.groupby('VendorID')['total_amount'].mean()
```

```python
#Filter the data for credit card payment only
credit_card = df[df['payment_type']==1]
print(credit_card)
```
